### Bivariate Analysis

A. Adjusting the 'sub_df' dataframe by __Data Aggregation__ and __Grouping techniques__.

### Data set 3 : Mean Sea Level Changes

```python
# Rename 'Measure' column to 'Water Regions'
sub_df.rename(columns={'Measure': 'Water Regions'}, inplace=True)

# Group by 'Water Regions' and 'Year', counting occurrences
grouped_df = sub_df.groupby(['Water Regions', 'Year']).agg(
    Count=('Value', 'size')  # Counts the number of occurrences
).reset_index()

# Group by 'Water Regions' and 'Year', calculating both count and average value
new_df = sub_df.groupby(['Water Regions', 'Year']).agg(
    Count=('Value', 'size'),          # Keeps the count from previous grouping
    Average_Value=('Value', 'mean')   # Calculates the mean of the 'Value'
).reset_index()

# Display the first 30 rows of the resultant dataset
new_df
```

```python
# Sort the data to put it in the right order for diff calculation
new_df.sort_values(by=['Water Regions', 'Year'], inplace=True)


# Define a function to categorize values
def categorize_value(x):
    if x > 0:
        return 'Increasing'
    elif x < 0:
        return 'Decreasing'
    else:
        return '0 level'

# Apply the function to the 'Average_Value' column to create a new column
new_df['Change Direction'] = new_df['Average_Value'].apply(categorize_value)


new_df.drop(columns=['Count'], inplace=True)


# Display the updated DataFrame
new_df
```

```python
# Function to categorize disaster frequency
def categorize_disaster_frequency(total):
    if total >= high_threshold:
        return 'High'
    elif total >= moderate_threshold:
        return 'Moderate'
    else:
        return 'Low'

# Filtering out the "Total" rows from the DataFrame
country_disasters = tidy_Disaster_freq[tidy_Disaster_freq['Disaster_type'] != 'TOTAL']

# Calculating total disaster frequency for each country
total_disasters_by_country = country_disasters.groupby('Country')['Frequency'].sum()

# Defining thresholds for disaster frequency categorization
high_threshold = total_disasters_by_country.quantile(0.75)
moderate_threshold = total_disasters_by_country.quantile(0.25)

# Categorizing countries based on total disaster frequency
disaster_prone_category = total_disasters_by_country.apply(categorize_disaster_frequency)

# Adding the new column 'Disaster_Prone_category' to the DataFrame
tidy_Disaster_freq['Disaster_Prone_category'] = tidy_Disaster_freq['Country'].map(disaster_prone_category)

#  updated DataFrame
tidy_Disaster_freq
```

Correlation measures the strength and direction of a linear relationship between two variables. It ranges from -1 to 1. A correlation of 1 indicates a perfect positive linear relationship, -1 indicates a perfect negative linear relationship, and 0 indicates no linear relationship.

- Here, the correlation value is __0.9015869000674023__, which is a positive number. 
- This means that as CO2 concentration increases, the surface temperature also tends to increase.
- As the correlation is very close to __1__, the relationship is high and almost perfectly linear.


```python
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Defining X and y values
X = merged_data[['Average CO2 Concentration']]
y = merged_data['Average Surface Temp']

# Defining a range of degrees for which we want to perform cross-validation
degrees = range(1, 6)

# Dictionary to store the cross-validation scores for each degree
cv_scores = {}

# Loop over the degrees
for degree in degrees:
    # Fitting a linear regression model for our polynomial by creating a pipeline
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression())
    
    # Performing cross-validation and storing the mean score in the dictionary
    scores = cross_val_score(model, X, y, cv=3)  # cv=5 is the number of folds
    cv_scores[degree] = scores.mean()

# Print the cross-validation scores
for degree, score in cv_scores.items():
    print(f"Degree: {degree}, Cross-validation score: {score}")
```

- __Degree 1__: The model with a polynomial of degree 1 has the highest score (close to zero), suggesting it generalizes the best out of the models tested.
- __Degree 2 to 5__: As the degree of the polynomial increases from 2 to 5, the score decreases (became negative), suggesting these models are overfitting the data and performing worse on the models tested.

```python

# Fitting a 1st degree polynomial to the data because as we have seen above Polynomial with degree 1 best fits
coefficients = np.polyfit(merged_data['Average CO2 Concentration'], merged_data['Average Surface Temp'], 1)
polynomial = np.poly1d(coefficients)

# Generating x values
x_values = np.linspace(merged_data['Average CO2 Concentration'].min(), merged_data['Average CO2 Concentration'].max(), 100)

# Generating y values
y_values = polynomial(x_values)

# scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Average CO2 Concentration', y='Average Surface Temp', data=merged_data)

# Plotting the fitted line
plt.plot(x_values, y_values, color='red')

#title and labels
plt.title('CO2 Concentration vs Surface Temperature')
plt.xlabel('Average CO2 Concentration')
plt.ylabel('Average Surface Temperature (in celsius)')

# Showing the plot
plt.show()
```
